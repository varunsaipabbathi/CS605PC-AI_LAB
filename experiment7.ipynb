```python
def monkeyboxproblem(state):
  print("initial state is:", state)
  state["monkey"] = state["box"]
  state["box"] = state["banana"]
  state["banana"] =state["box"]
  print("goal state is:" ,state)

one=input("enter the monkey position:")
two=input("enter the box position:")
three=input("enter the banana position:")
state = {"monkey": one, "box": two, "banana": three}
monkeyboxproblem(state)

```
